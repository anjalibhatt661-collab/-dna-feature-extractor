```python
import os

drive_data_path = '/content/drive/MyDrive/skin-lesion-classifier-data'

# Check counts in both casing versions
img_dir1 = os.path.join(drive_data_path, 'HAM10000_images_part_1')
img_dir2 = os.path.join(drive_data_path, 'HAM10000_images_part_2')

count1 = len(os.listdir(img_dir1))
count2 = len(os.listdir(img_dir2))

print(f"Part 1: {count1} images")
print(f"Part 2: {count2} images")
print(f"Total: {count1 + count2}")  # expect 10015
```

```python
import shutil

# Remove duplicate lowercase image folders
shutil.rmtree(os.path.join(drive_data_path, 'ham10000_images_part_1'))
shutil.rmtree(os.path.join(drive_data_path, 'ham10000_images_part_2'))

# Remove unused pre-flattened pixel CSVs
for f in ['hmnist_8_8_L.csv', 'hmnist_8_8_RGB.csv', 'hmnist_28_28_L.csv', 'hmnist_28_28_RGB.csv']:
    path = os.path.join(drive_data_path, f)
    if os.path.exists(path):
        os.remove(path)

print("Cleanup done.")
print(os.listdir(drive_data_path))
```
